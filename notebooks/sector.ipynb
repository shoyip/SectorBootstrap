```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt

from cocoatree.pysca import _icList
```

```python
mpl.rcParams['text.usetex'] = False
mpl.rcParams['axes.titlesize'] = 20
mpl.rcParams['axes.labelsize'] = 18
mpl.rcParams['lines.linewidth'] = 1
plt.rcParams['xtick.labelsize']=15
plt.rcParams['ytick.labelsize']=15
```

```python
eigvals = np.load("../data/red_sector_eigvals.npy", allow_pickle=True)
```

```python
residues = np.load("../data/red_sector_indices.npy", allow_pickle=True)
```

```python
print(residues, len(residues))
```

```python
plt.figure()
plt.title("Eigenvalues on the red sector component")
plt.bar(np.arange(len(eigvals)), eigvals)
plt.xlabel("Residue index")
plt.ylabel("Eigenvalue on component")
plt.show()
```

```python
plt.figure()
plt.title("Ordered values of eigenvalues on red sector")
plt.plot(np.sort(np.abs(eigvals))[::-1], label="Values of eigenvalues on red sector component")
plt.axhline(eigvals[residues[-1]], linestyle='--', color='tab:red', label="Red sector threshold")
plt.legend()
plt.xlabel("Rank")
plt.ylabel("Eigenvalue on component")
plt.show()
```
